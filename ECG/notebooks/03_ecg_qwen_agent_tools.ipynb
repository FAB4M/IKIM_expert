```python
import neurokit2 as nk
import numpy as np

def run_rhythm_feature_tool(record_path):
    signal, fields = wfdb.rdsamp(str(record_path))
    fs = int(fields["fs"])
    lead_names = fields["sig_name"]

    if "II" in lead_names:
        lead_idx = lead_names.index("II")
        lead_used = "II"
    else:
        lead_idx = 0
        lead_used = lead_names[0]

    ecg_lead = signal[:, lead_idx]

    try:
        signals, info = nk.ecg_process(ecg_lead, sampling_rate=fs)
        rpeaks = info.get("ECG_R_Peaks", [])

        if len(rpeaks) >= 2:
            rr_intervals_sec = np.diff(rpeaks) / fs
            mean_rr = float(np.mean(rr_intervals_sec))
            sd_rr = float(np.std(rr_intervals_sec))
            cv_rr = float(sd_rr / mean_rr) if mean_rr > 0 else np.nan
            estimated_hr = float(60 / mean_rr) if mean_rr > 0 else np.nan

            if cv_rr < 0.10:
                rhythm_regularity = "likely regular"
            elif cv_rr < 0.20:
                rhythm_regularity = "mildly irregular"
            else:
                rhythm_regularity = "irregular"
        else:
            mean_rr = np.nan
            sd_rr = np.nan
            cv_rr = np.nan
            estimated_hr = np.nan
            rhythm_regularity = "insufficient R-peaks detected"

        return {
            "tool_name": "NeuroKit2 rhythm feature tool",
            "lead_used": lead_used,
            "sampling_rate_hz": fs,
            "number_of_r_peaks": int(len(rpeaks)),
            "estimated_heart_rate_bpm": estimated_hr,
            "mean_rr_sec": mean_rr,
            "sd_rr_sec": sd_rr,
            "rr_coefficient_of_variation": cv_rr,
            "rhythm_regularity": rhythm_regularity,
            "interpretation_scope": "Rhythm feature assessment only; not a full arrhythmia diagnosis."
        }

    except Exception as e:
        return {
            "tool_name": "NeuroKit2 rhythm feature tool",
            "lead_used": lead_used,
            "sampling_rate_hz": fs,
            "error": str(e),
            "interpretation_scope": "Failed rhythm feature extraction."
        }
```

```python
import os
os.environ["KERAS_BACKEND"] = "tensorflow"

from huggingface_hub import hf_hub_download
import keras
import numpy as np
import json

repo_id = "Steenslid/ecg-ptbxl-classification"

model_path = hf_hub_download(
    repo_id=repo_id,
    filename="ecg_model.keras"
)

norm_path = hf_hub_download(
    repo_id=repo_id,
    filename="normalisation_params.npz"
)

thresholds_path = hf_hub_download(
    repo_id=repo_id,
    filename="thresholds.json"
)

print("Model path:", model_path)
print("Normalisation path:", norm_path)
print("Thresholds path:", thresholds_path)

ptbxl_model = keras.saving.load_model(model_path)

normalisation_params = np.load(norm_path)

with open(thresholds_path, "r") as f:
    thresholds = json.load(f)

print("Loaded pretrained PTB-XL classifier.")
print("Model input shape:", ptbxl_model.input_shape)
print("Model output shape:", ptbxl_model.output_shape)
print("Normalisation keys:", normalisation_params.files)
print("Thresholds:", thresholds)
```

```python
def normalize_ecg_signal(x, normalisation_params):
    """
    x shape: 1000 x 12
    """
    keys = normalisation_params.files

    if "mean" in keys and "std" in keys:
        mean = normalisation_params["mean"]
        std = normalisation_params["std"]
        return (x - mean) / (std + 1e-8)

    elif "means" in keys and "stds" in keys:
        mean = normalisation_params["means"]
        std = normalisation_params["stds"]
        return (x - mean) / (std + 1e-8)

    elif "mu" in keys and "sigma" in keys:
        mean = normalisation_params["mu"]
        std = normalisation_params["sigma"]
        return (x - mean) / (std + 1e-8)

    else:
        print("Warning: Unknown normalisation keys. Returning unnormalized signal.")
        return x


def run_ptbxl_classifier_tool(record_path):
    signal, fields = wfdb.rdsamp(str(record_path))

    x = signal.astype("float32")

    # Erwartet: 1000 x 12
    if x.shape[0] != 1000:
        if x.shape[0] > 1000:
            x = x[:1000, :]
        else:
            pad_len = 1000 - x.shape[0]
            x = np.pad(x, ((0, pad_len), (0, 0)), mode="constant")

    x = normalize_ecg_signal(x, normalisation_params)

    x_batch = np.expand_dims(x, axis=0)

    y_pred = ptbxl_model.predict(x_batch, verbose=0)

    if isinstance(y_pred, list):
        y_pred = y_pred[0]

    probs = np.array(y_pred).reshape(-1)

    labels = ["NORM", "MI", "STTC", "CD", "HYP"]

    # thresholds.json kann entweder Liste oder Dict sein
    if isinstance(thresholds, dict):
        predicted_labels = [
            label for label, prob in zip(labels, probs)
            if prob >= float(thresholds.get(label, 0.5))
        ]
    else:
        predicted_labels = [
            label for label, prob, thr in zip(labels, probs, thresholds)
            if prob >= float(thr)
        ]

    if len(predicted_labels) == 0:
        predicted_labels = [labels[int(np.argmax(probs))]]

    return {
        "tool_name": "pretrained PTB-XL classifier",
        "classes": labels,
        "probabilities": {label: float(prob) for label, prob in zip(labels, probs)},
        "predicted_labels": predicted_labels,
        "top_class": labels[int(np.argmax(probs))],
        "interpretation_scope": "Pretrained signal-based PTB-XL superclass classifier."
    }
```

```python
def normalize_ecg_signal(x, normalisation_params):
    """
    x shape expected: 1000 x 12
    Returns shape: 1000 x 12
    """
    keys = normalisation_params.files

    if "mean" in keys and "std" in keys:
        mean = normalisation_params["mean"]
        std = normalisation_params["std"]
    elif "means" in keys and "stds" in keys:
        mean = normalisation_params["means"]
        std = normalisation_params["stds"]
    elif "mu" in keys and "sigma" in keys:
        mean = normalisation_params["mu"]
        std = normalisation_params["sigma"]
    else:
        print("Warning: Unknown normalisation keys. Returning unnormalized signal.")
        return x

    # Entferne unnötige Batch-Dimensionen, falls vorhanden
    mean = np.squeeze(mean)
    std = np.squeeze(std)

    # Falls mean/std pro Lead vorliegen, z.B. shape (12,)
    # oder vollständig pro Sample/Lead, z.B. shape (1000, 12), funktioniert Broadcasting.
    x_norm = (x - mean) / (std + 1e-8)

    # Sicherstellen, dass Ergebnis wieder 1000 x 12 ist
    x_norm = np.squeeze(x_norm)

    if x_norm.shape != (1000, 12):
        print("Warning: normalized signal has unexpected shape:", x_norm.shape)

    return x_norm.astype("float32")


def run_ptbxl_classifier_tool(record_path):
    signal, fields = wfdb.rdsamp(str(record_path))

    x = signal.astype("float32")

    # Erwartet: 1000 x 12
    if x.shape[0] != 1000:
        if x.shape[0] > 1000:
            x = x[:1000, :]
        else:
            pad_len = 1000 - x.shape[0]
            x = np.pad(x, ((0, pad_len), (0, 0)), mode="constant")

    # Falls mehr als 12 Kanäle oder weniger, prüfen
    if x.shape[1] != 12:
        raise ValueError(f"Expected 12 leads, but got shape {x.shape}")

    print("Raw signal shape:", x.shape)

    x = normalize_ecg_signal(x, normalisation_params)

    print("Normalized signal shape:", x.shape)

    # Modell erwartet Batch x 1000 x 12
    x_batch = np.expand_dims(x, axis=0)

    print("Model input shape:", x_batch.shape)

    y_pred = ptbxl_model.predict(x_batch, verbose=0)

    if isinstance(y_pred, list):
        y_pred = y_pred[0]

    probs = np.array(y_pred).reshape(-1)

    labels = ["NORM", "MI", "STTC", "CD", "HYP"]

    # thresholds.json kann entweder Liste oder Dict sein
    if isinstance(thresholds, dict):
        predicted_labels = [
            label for label, prob in zip(labels, probs)
            if prob >= float(thresholds.get(label, 0.5))
        ]
    else:
        predicted_labels = [
            label for label, prob, thr in zip(labels, probs, thresholds)
            if prob >= float(thr)
        ]

    if len(predicted_labels) == 0:
        predicted_labels = [labels[int(np.argmax(probs))]]

    return {
        "tool_name": "pretrained PTB-XL classifier",
        "classes": labels,
        "probabilities": {label: float(prob) for label, prob in zip(labels, probs)},
        "predicted_labels": predicted_labels,
        "top_class": labels[int(np.argmax(probs))],
        "interpretation_scope": "Pretrained signal-based PTB-XL superclass classifier."
    }
```

```python
LABEL_MEANINGS = {
    "NORM": "normal ECG",
    "MI": "myocardial infarction",
    "STTC": "ST/T changes",
    "CD": "conduction disturbance",
    "HYP": "hypertrophy"
}

def create_agent_summary_from_tools(rhythm_output, ptbxl_output):
    probs = ptbxl_output["probabilities"]
    sorted_probs = sorted(probs.items(), key=lambda x: x[1], reverse=True)

    predicted_labels = ptbxl_output.get("predicted_labels", [])
    top_label, top_prob = sorted_probs[0]
    top_meaning = LABEL_MEANINGS.get(top_label, top_label)

    lines = []
    lines.append("ECG Agent Summary")
    lines.append("")
    lines.append("1. Rhythm assessment")
    lines.append(f"- Lead used: {rhythm_output.get('lead_used', 'unknown')}")
    lines.append(f"- Number of detected R-peaks: {rhythm_output.get('number_of_r_peaks', 'unknown')}")

    hr = rhythm_output.get("estimated_heart_rate_bpm", None)
    if isinstance(hr, (int, float)) and not np.isnan(hr):
        lines.append(f"- Estimated heart rate: {hr:.1f} bpm")
    else:
        lines.append("- Estimated heart rate: unavailable")

    lines.append(f"- Rhythm regularity: {rhythm_output.get('rhythm_regularity', 'unknown')}")
    lines.append("- Scope: rhythm feature extraction only; not full arrhythmia diagnosis.")
    lines.append("")
    lines.append("2. Diagnostic superclass classification")
    lines.append(f"- Predicted label(s) above threshold: {', '.join(predicted_labels)}")
    lines.append(f"- Highest probability class: {top_label} ({top_meaning}), probability {top_prob:.3f}")
    lines.append("")
    lines.append("Class probabilities:")
    for label, prob in sorted_probs:
        meaning = LABEL_MEANINGS.get(label, label)
        lines.append(f"- {label} ({meaning}): {prob:.3f}")

    lines.append("")
    lines.append("3. Integrated interpretation")

    if "NORM" in predicted_labels and len(predicted_labels) == 1:
        lines.append("- The pretrained signal classifier primarily suggests a normal ECG pattern.")
    elif len(predicted_labels) > 0:
        expanded = [f"{label} ({LABEL_MEANINGS.get(label, label)})" for label in predicted_labels]
        lines.append(
            f"- The pretrained signal classifier suggests abnormal PTB-XL superclass label(s): {', '.join(expanded)}."
        )
    else:
        lines.append(
            f"- No class exceeded the predefined threshold; the highest probability class was {top_label} ({top_meaning})."
        )

    lines.append("- The rhythm feature tool provides supportive rhythm information based on R-peak detection.")
    lines.append("- This is a research prototype and not a clinical diagnosis.")

    return "\n".join(lines)
```

```python
def run_ecg_agent(row=None, record_id=None):
    if row is None:
        if record_id is None:
            raise ValueError("Please provide either row or record_id.")
        row = df_agent[df_agent["ecg_id"].astype(int) == int(record_id)].iloc[0]

    record_path = PTBXL_DIR / row["filename_lr"]

    rhythm_output = run_rhythm_feature_tool(record_path)
    ptbxl_output = run_ptbxl_classifier_tool(record_path)
    agent_summary = create_agent_summary_from_tools(rhythm_output, ptbxl_output)

    return {
        "ecg_id": int(row["ecg_id"]),
        "record_path": str(record_path),
        "true_diagnostic_class_main": row.get("diagnostic_class_main", None),
        "true_diagnostic_classes": row.get("diagnostic_classes", None),
        "rhythm_tool_output": rhythm_output,
        "ptbxl_classifier_output": ptbxl_output,
        "agent_summary": agent_summary
    }
```

```python
import os
os.environ["KERAS_BACKEND"] = "tensorflow"

import json
import numpy as np
import pandas as pd
import wfdb
import keras
from pathlib import Path
from huggingface_hub import hf_hub_download

PROJECT_DIR = Path("/content/drive/MyDrive/ecg-qwen-modality-expert")
DATA_DIR = PROJECT_DIR / "data"
RESULTS_DIR = PROJECT_DIR / "results"
PTBXL_DIR = DATA_DIR / "ptbxl_mini"

# Falls df_agent nicht mehr existiert, neu laden
if "df_agent" not in globals():
    candidate_paths = [
        RESULTS_DIR / "ecg_qa_training_table_2500_with_images.csv",
        RESULTS_DIR / "ecg_qa_training_table_1000_with_images.csv",
        RESULTS_DIR / "mini_ecg_qa_training_table_with_images.csv",
    ]

    existing_candidates = [p for p in candidate_paths if p.exists()]

    if len(existing_candidates) == 0:
        found = list(Path("/content/drive/MyDrive").rglob("*training_table*with_images*.csv"))
        if len(found) == 0:
            raise FileNotFoundError("No ECG training table found.")
        csv_path = found[0]
    else:
        csv_path = existing_candidates[0]

    df_agent = pd.read_csv(csv_path)
    print("Loaded df_agent from:", csv_path)

print("df_agent shape:", df_agent.shape)

# Externes Modell von Hugging Face laden
repo_id = "Steenslid/ecg-ptbxl-classification"

model_path = hf_hub_download(
    repo_id=repo_id,
    filename="ecg_model.keras"
)

norm_path = hf_hub_download(
    repo_id=repo_id,
    filename="normalisation_params.npz"
)

thresholds_path = hf_hub_download(
    repo_id=repo_id,
    filename="thresholds.json"
)

ptbxl_model = keras.saving.load_model(model_path)
normalisation_params = np.load(norm_path)

with open(thresholds_path, "r") as f:
    thresholds = json.load(f)

print("Loaded pretrained PTB-XL classifier.")
print("Model input shape:", ptbxl_model.input_shape)
print("Model output shape:", ptbxl_model.output_shape)
print("Normalisation keys:", normalisation_params.files)
print("Thresholds:", thresholds)


def normalize_ecg_signal(x, normalisation_params):
    """
    x expected shape: 1000 x 12
    returns: 1000 x 12
    """
    keys = normalisation_params.files

    if "mean" in keys and "std" in keys:
        mean = normalisation_params["mean"]
        std = normalisation_params["std"]
    elif "means" in keys and "stds" in keys:
        mean = normalisation_params["means"]
        std = normalisation_params["stds"]
    elif "mu" in keys and "sigma" in keys:
        mean = normalisation_params["mu"]
        std = normalisation_params["sigma"]
    else:
        print("Warning: Unknown normalisation keys. Returning unnormalized signal.")
        return x.astype("float32")

    mean = np.squeeze(mean)
    std = np.squeeze(std)

    x_norm = (x - mean) / (std + 1e-8)
    x_norm = np.squeeze(x_norm)

    return x_norm.astype("float32")


def run_ptbxl_classifier_tool(record_path):
    signal, fields = wfdb.rdsamp(str(record_path))

    x = signal.astype("float32")

    # Erwartet: 1000 x 12
    if x.shape[0] != 1000:
        if x.shape[0] > 1000:
            x = x[:1000, :]
        else:
            pad_len = 1000 - x.shape[0]
            x = np.pad(x, ((0, pad_len), (0, 0)), mode="constant")

    if x.shape[1] != 12:
        raise ValueError(f"Expected 12 leads, but got shape {x.shape}")

    x = normalize_ecg_signal(x, normalisation_params)

    # Modell erwartet: batch x 1000 x 12
    x_batch = np.expand_dims(x, axis=0)

    y_pred = ptbxl_model.predict(x_batch, verbose=0)

    if isinstance(y_pred, list):
        y_pred = y_pred[0]

    probs = np.array(y_pred).reshape(-1)

    labels = ["NORM", "MI", "STTC", "CD", "HYP"]

    if isinstance(thresholds, dict):
        predicted_labels = [
            label for label, prob in zip(labels, probs)
            if prob >= float(thresholds.get(label, 0.5))
        ]
    else:
        predicted_labels = [
            label for label, prob, thr in zip(labels, probs, thresholds)
            if prob >= float(thr)
        ]

    if len(predicted_labels) == 0:
        predicted_labels = [labels[int(np.argmax(probs))]]

    return {
        "tool_name": "pretrained PTB-XL classifier",
        "classes": labels,
        "probabilities": {label: float(prob) for label, prob in zip(labels, probs)},
        "predicted_labels": predicted_labels,
        "top_class": labels[int(np.argmax(probs))],
        "interpretation_scope": "Pretrained signal-based PTB-XL superclass classifier."
    }

print("run_ptbxl_classifier_tool is ready.")
```
